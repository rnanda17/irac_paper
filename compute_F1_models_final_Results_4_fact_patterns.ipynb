```python
import pandas as pd

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", None)

df = pd.read_csv("metrics_all_LLMs.csv")

# Show Q2 rows for manual verification
print("=== Q2 rows being removed ===")
df[df["question_number"] == "Q2"]

```

```python
# Remove Q2 and compute macro averages
df_no_q2 = df[df["question_number"] != "Q2"]
df_no_q2
```

```python


df_no_q2.groupby(["model", "metric"])[["precision", "recall", "f1"]].mean().round(3).reset_index().sort_values(["metric", "f1"], ascending=[True, False])

```
